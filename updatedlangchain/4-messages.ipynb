### Messages

Message are the fundamental unit of context for models in LangChain. They represent the input and output of the models, carrying both the content and metadata needed to represent the state of a converstation when interacting with a LLM. Message are objects that contain"

1. Role - identifies the message type(e.g system, user)
2. Content - Representes the actual content of the message (like text, images, audio, documents, etc.)
3. Metadata - Optional fields such as response information, message IDs, and token usage

LangChain porvides a standard message type theat works across all model providers, ensuring consistent behavior regardless of the model being called.

### Text Prompts

Text prompts are strings - ideal for straight forward generation tasks where you don't need to retain conversation history.

```python
model.invoke("What is lagchain")
```

Use text prompts when:

1. You have a single, standalone request
2. You don't need conversation hsitory
3. You want minimal code complexity

### Message Prompts ###

Alternatively, you can pass in a list messages to the model by providing a list of message objects.

Message types:

1) System message - Tells the model how to behave and provide context for interactions
2) Human message - Represents user input and interactions with the model
3) AI message - Reponsees generated by the model, including text content, tool calls, and metadata
4) Tool message - Represents the outputs of tool calls

### System Message ###

A System Message represent an initial set of instruction that primes the model's behavior. You can use a system message to set the tone. define the model's role, and establish guideline for responses.

### Human Message ###

A Human Message represent user input and interactions. They can contain text, images, audio, files and anyother amount of multimodal content.

### AI Messages ###

An AI Message represents the output of a model invocation. They can include multimodal data, tool calls, and provider-sepcific metadata that you can later access.

### Tool Message ###

For models that support tool calling. AI messages can contain tool calls. Tool messages are used to pass the results of a single tool execuation back to the model.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("Your are a poetry expert"),
    HumanMessage("Write a poem on artifical intelligence")
]

response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are a helpful coding assistant")

messages = [
    system_msg,
    HumanMessage("How do i create a REST API")
]

response = model.invoke(messages)
print(response.content)
```

```python
## Detailed info to the LLM through System message

from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior python developer with expertise in web frameworks.
Always provide code examples and explain your reasoning.
Be concise but through in your explanation.
""")

message = [
    system_msg,
    HumanMessage("How do i create a REST API?")
]

response = model.invoke(message)
print(response.content)
```

```python
## Message Metadata

human_msg = HumanMessage(
    content="Hello!",
    # The below key value pair is called metadata.
    name="alice", # Optional: identify different users
    id="msg_123", # Optional: unique identifier for tracing
)

response = model.invoke([human_msg])
response
```

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage

ai_msg = AIMessage("i`d be happy to help you with that question!")

# Add to conversation History
messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("Can you help me?"),
    ai_msg,
    HumanMessage("Great! What's 2+2")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes a tool call
# (Here, we demonstrate manually creating the messages for brevity)

ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Franciso"},
        "id": "call_123"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 72 Degree F"
tool_message = ToolMessage(
    content = weather_result,
    tool_call_id = "call_123" # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message, # Model's tool call
    tool_message, # Tool Execution result
]
```
