### Streaming and Batch

### Streaming

Most models can stream their output content while it is being generated.By displaying output progressively, streaming significantly imporves user expreience, particularly for longer response. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in reat-time.

```python
# in here we need to wait untill the llm give response 
model.invoke("Write me a 200 words paragraph on artificial intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on artificial intelligence"):
    print(chunk.text, end= " | ", flush=True) 
```

### BATCH

Batching a collection of independent requrets to a model can significantly imporve performance and reduce costs, as the proecessing can be done in parallel;

```python
responses = model.batch(
    [
        "Why do parrots have colorful feathres?",
        "How do airplanes fly?",
        "What is quantum computing?"
    ]
)

for response in responses:
    print(response)
```

```python
responses = model.batch(
    [
        "Why do parrots have colorful feathres?",
        "How do airplanes fly?",
        "What is quantum computing?", 
    ],
    config={
        'max_concurrency': 5 # Limit to 5 parallel calls
    }
)


for response in responses:
    print(response)
```
